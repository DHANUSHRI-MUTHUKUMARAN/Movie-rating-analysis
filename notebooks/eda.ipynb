```python
import pandas as pd

movies = pd.read_csv("../data/raw/movies.csv")
ratings = pd.read_csv("../data/raw/ratings.csv")
tags = pd.read_csv("../data/raw/tags.csv")
links = pd.read_csv("../data/raw/links.csv")

print("Movies:", movies.shape)
print("Ratings:", ratings.shape)
print("Tags:", tags.shape)
print("Links:", links.shape)
```

```python
print("Movies")
print(movies.isnull().sum())

print("\nRatings")
print(ratings.isnull().sum())

print("\nTags")
print(tags.isnull().sum())
```

```python
print("Movies:", movies.duplicated().sum())
print("Ratings:", ratings.duplicated().sum())
print("Tags:", tags.duplicated().sum())
```

```python
df = ratings.merge(movies, on="movieId", how="left")

df.head()
```

```python
print("Total Ratings:", len(df))
print("Unique Users:", df["userId"].nunique())
print("Unique Movies:", df["movieId"].nunique())
print("Average Rating:", round(df["rating"].mean(), 2))
print("Minimum Rating:", df["rating"].min())
print("Maximum Rating:", df["rating"].max())
```

```python
df.to_csv("../data/processed/merged_movies.csv", index=False)

print("Merged dataset saved successfully!")
```
